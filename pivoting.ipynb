Pivoting

```python
# Create Dataframe for monthly sales 
# Pivot month as rows and products as column

sales = pd.DataFrame({
  "month":["January" , "January",  "February",  "February" , "March" , "March"],
  "product":["Laptop" , "Mobile" , "Laptop" , "Mobile" , "Laptop" , "Mobile"],
  "sales":[50000 , 30000 , 55000 , 35000 , 60000 , 40000]
})

print(sales)
```

```python
#pivot() 
#index -> rows
#columns -> columns

month_order = ["January", "February", "March"]
# month_order = ["March"]

sales ["month"] = pd.Categorical(sales["month"], categories=month_order, ordered=True)

sales_pivot = sales.pivot(index= "month", columns="product", values="sales")

#sales_pivot = sales_pivot.sort_index()
print(sales_pivot.loc["March"])
print("===============================")
print(sales_pivot)
```
