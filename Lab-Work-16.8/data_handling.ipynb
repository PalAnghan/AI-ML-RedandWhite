Topic: Categorical Data Handling


Task 1: Handle Categorical Variables

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
print("Create a DataFrame with a 'Color' column having repeated values.")

df = pd.DataFrame({
  "Name":["Rahul" , "Amit" , "Zeel" , "Riya" , "Priya" , "Pooja" , "Neha" , "Karan"],
  "Color":["Red" , "Blue",  "Red" , "Green" , "Red" , "Blue" , "Green" , "Red"]
})

print(df)
```

```python
# Important DataFrame Attribute

print("shape: ",df.shape)
print("Columns: ",df.columns)
print("Index: ",df.index)
print("DataTypes: ",df.dtypes)
print("Values: ",df.values)
print("Size: ",df.size)
print("ndim: ",df.ndim)
```

Convert the 'Color' column into a pandas Categorical type

```python
df["Color"] = df["Color"].astype("category")

print(df)

print(df["Color"].dtype)
```

```python
print("Count the number of unique categories in the 'Color' column.")


color_category = pd.Categorical(
  [
    "Red",
    "Blue",
    "Red",
    "Green"
  ]
)

print(color_category)
```

Task 2: Perform Categorical Analysis

```python
unique_count = df["Color"].nunique()
# unique_count = df["Color"].unique()

print(unique_count)
```

```python
print("Display the frequency of each category.")

color_frequency = df["Color"].value_counts()

print(color_frequency)
```

```python
# percentage = df["Color"].value_counts(normalize=True) * 100
percentage = df["Color"].value_counts(ascending=False) * 100

print(percentage)
```

Task 3: Compare and Order Categories

```python
print(" Create a 'Size' column with values: 'Small', 'Medium', 'Large'.")

df["Size"] = ["Medium" , "Small" , "Large" , "Medium" , "Small" , "Large" , "Small" , "Large"]

print(df)
```

```python
print(" Define a logical order for these categories.")

size_order = ["Small" , "Medium" , "Large"]

df["Size"] = pd.Categorical(
  df["Size"],
  categories = size_order
)

df_sorted = df.sort_values("Size")

print(df_sorted)
print(df)
```

Task 4: Rename Categories

```python
print(" Rename the categories in the 'Size' column to 'S', 'M', and 'L'.")

print(df["Size"].cat.categories)
```

```python
df["Size"] = df["Size"].cat.rename_categories({
  "Small" : "S",
  "Medium": "M",
  "Large" : "L"
})

print(df.dtypes)
print(df)
```

```python
print("BAR CHART")

size_count = df["Size"].value_counts()

print(size_count)
```

```python
plt.figure(figsize=(5 , 5))
size_count.plot(kind="bar", color="pink")
plt.title("Number of Students by Size")
plt.xlabel("Size")
plt.ylabel("Count")
plt.show()
```
